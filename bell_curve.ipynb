```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

marks = [40,30,56,54,51,55,48,52,48,49,49,30,53,39,41,45,42,30,30,52,
         32,50,30,38,48,49,49,52,45,52,56,53,51,52,46,49,47,46,48,43,
         48,46,39,47,52,40,52,51,52,52,45,41,49,39,36,30,44,56,44,45,
         43,51,44,37]

mu, sigma = np.mean(marks), np.std(marks)
x = np.linspace(25, 60, 200)

plt.hist(marks, bins=range(30, 62, 5), density=True, edgecolor='black', alpha=0.6)
plt.plot(x, norm.pdf(x, mu, sigma), 'r', lw=2)
plt.axvline(mu, color='k', linestyle='--')
plt.title(f'Mark Distribution (mean={mu:.1f}, SD={sigma:.1f})')
plt.xlabel('Marks'); plt.ylabel('Density')
plt.show()
```
