```python
import pandas as pd

df = pd.DataFrame({
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Marks": [45, 48, 50, 52, 49, 51, 47, 95]
})

print(df)
```

```python
df['Marks'].mean()
df['Marks'].median()
```

```python
df.describe()
```

```python
q1 = df['Marks'].quantile(0.25)
q2 = df['Marks'].quantile(0.50)
q3 = df['Marks'].quantile(0.75)
iqr = q3 - q1
print(q1)
print(q2)
print(q3)
print(iqr)
```

```python
lower=q1-1.5*iqr
upper=q3+1.5*iqr
print(lower)
print(upper)
```

```python
outliers = df[(df['Marks'] < lower) | (df['Marks'] > upper)]
print(f"Lower Bound: {lower}")
print(f"Upper Bound: {upper}\n")
print("Outliers:")
print(outliers)
```

```python
import pandas as pd
data = {
    'Group_A': [45, 48, 50, 52, 49, 51, 47, 50],
    'Group_B': [60, 62, 65, 63, 61, 64, 66, 95]
}

df = pd.DataFrame(data)
def analyze_data(series):
    mean_val = series.mean()
    median_val = series.median()
    min_val = series.min()
    max_val = series.max()
    std_val = series.std()
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    outliers = series[(series < lower_bound) | (series > upper_bound)].tolist()

    return {
        'Mean': mean_val,
        'Median': median_val,
        'Min': min_val,
        'Max': max_val,
        'Std Dev': round(std_val, 2),
        'Q1': q1,
        'Q3': q3,
        'IQR': iqr,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Outliers': outliers if outliers else "None"
    }
results = pd.DataFrame({
    'Group_A': analyze_data(df['Group_A']),
    'Group_B': analyze_data(df['Group_B'])
})

print(results)
```
